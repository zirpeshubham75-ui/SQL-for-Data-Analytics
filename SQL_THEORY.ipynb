# 7 DDL Commands

DDL stands for Data Definition Language.
DDL commands are used to create, modify, and delete the structure of database objects, such as tables, databases, columns, etc.

The main DDL commands are:

CREATE
ALTER
DROP
TRUNCATE

# 1. CREATE

Interview definition:
CREATE is a DDL command used to create a new database object, such as a database or table.

# 2. ALTER

Interview definition:
ALTER is used to modify the existing structure of a database object, mainly a table.

We can use it to:

Add a column
Modify a column
Rename a column
Drop a column

# 3. DROP

Interview definition:
DROP is used to permanently delete a database object

# 4. TRUNCATE

Interview definition:
TRUNCATE is used to remove all records from a table while keeping the table structure.

# 8 .  DELETE VS  TRUNCATE   VS DROP

# 10.DML Commands — Interview Ready

DML stands for Data Manipulation Language.
DML commands are used to insert, modify, and delete data stored inside a table.

The main DML commands are:

INSERT  - “INSERT is a DML command used to add new records or rows into a table.”
UPDATE  -  “UPDATE is a DML command used to modify or change existing records in a table.”
DELETE   - DELETE is a DML command used to remove existing records or rows from a table.

# 11 . What are Constraints?

Interview answer:

“Constraints are rules applied to columns in a table to maintain data accuracy, consistency,


| Constraint      | Interview Definition                                                         | Example                                 |
| --------------- | ---------------------------------------------------------------------------- | --------------------------------------- |
| **NOT NULL**    | Ensures a column cannot contain `NULL` values.                               | `name VARCHAR(50) NOT NULL`             |
| **UNIQUE**      | Ensures all values in a column are different.                                | `email VARCHAR(100) UNIQUE`             |
| **PRIMARY KEY** | Uniquely identifies each record in a table. It cannot contain `NULL` values. | `id INT PRIMARY KEY`                    |
| **FOREIGN KEY** | Creates a relationship between two tables and ensures referential integrity. | `dept_id INT REFERENCES department(id)` |
| **CHECK**       | Ensures values satisfy a specified condition.                                | `age INT CHECK(age >= 18)`              |
| **DEFAULT**     | Provides a default value when no value is supplied.                          | `city VARCHAR(50) DEFAULT 'Pune'`       |
